```python
#Load spaCy and Process the Text
import spacy
# Input sentence
text ="The intelligent student quickly completed her assignment and" \
" submitted  it to the teacher. She  was very  happy because" \
" she had worked hard  on the project. " \
"The teacher praised her excellent  work and gave her a certificate. " \
"She sat in the classroom and thanked her teacher with  a smile." \
" Wow! she felt proud of her achievement."\

# Load the English language model
nlp = spacy.load("en_core_web_sm")
# Process the sentence
doc = nlp(text)

print("Text processed successfully.")
print("Number of tokens:", len(doc))

```

```python
# Display POS Tags

# Print each word and its Part of Speech
for token in doc:
    print(token.text, "->", token.pos_)
```

```python
#Display POS in a Table
import pandas as pd

pos_table = pd.DataFrame({
"Token": [token.text for token in doc],
"POS": [token.pos_ for token in doc]
})

print(pos_table.to_string(index=False))
```

```python
#Display Detailed POS Tags
# POS gives a general category, while Tag gives more detailed
#  grammatical information from the English model.
detailed_table = pd.DataFrame({
"Token": [token.text for token in doc],
"POS": [token.pos_ for token in doc],
"Tag": [token.tag_ for token in doc]
})

print(detailed_table.to_string(index=False))
```

```python
#Display POS, Lemma and Morphology Together 
# The lemma gives the base form of a word,
# Morphological describe information such as
# tense, number, person or degree

analysis_table = pd.DataFrame({
"Token": [token.text for token in doc],
"POS": [token.pos_ for token in doc],
"Tag": [token.tag_ for token in doc],
"Lemma": [token.lemma_ for token in doc],
"Morphology": [str(token.morph) for token in doc]
})

print(analysis_table.to_string(index=False))
```

```python
# Analyze Only Alphabetic Words
word_table = pd.DataFrame({
"Token": [t.text for t in doc if t.is_alpha],
"POS": [t.pos_ for t in doc if t.is_alpha],
"Tag": [t.tag_ for t in doc if t.is_alpha],
"Lemma": [t.lemma_ for t in doc if t.is_alpha]
})

print(word_table.to_string(index=False))
```

```python
#Count the Frequency of Each POS Category
pos_counts = (
word_table["POS"]
.value_counts()
.rename_axis("POS")
.reset_index(name="Count")
)

print(pos_counts)
```

```python
#Optional Visualization
import matplotlib.pyplot as plt

pos_counts.plot(
kind="bar",
x="POS",
y="Count",
legend=False,
title="POS Tag Distribution"
)

plt.xlabel("Part of Speech")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

```python



# List of sentences
sentences = [
    "I read a book every week.",
    "Please book a ticket for tomorrow.",
    "They watch the match.",
    "They match the colors carefully.",
    "The light is bright.",
    "Please light the lamp."
]

# Process every sentence
for sentence in sentences:
    print("\nSentence:", sentence)

    # Analyze the current sentence
    sentence_doc = nlp(sentence)

    # Check each word
    for token in sentence_doc:
        if token.text.lower() in {"book", "match", "light"}:
            print(
                f"Word: {token.text:10} "
                f"POS={token.pos_:8} "
                f"TAG={token.tag_:8} "
                f"LEMMA={token.lemma_}"
            )
```
