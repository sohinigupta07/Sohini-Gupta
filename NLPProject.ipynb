```python
#Tokenization
import spacy
text = "Natural language processing helps computers understand human language."

nlp = spacy.load("en_core_web_sm")
doc = nlp(text)

print("Number of spaCy tokens:", len(doc))
print(text)
```

```python
#Whitespace tokenization

whitespace_tokens = text.split()

print("Whitespace tokens:")
print(whitespace_tokens)
print("Count:", len(whitespace_tokens))
```

```python
#spaCy word tokenization

spacy_tokens = [token.text for token in doc]

print("spaCy tokens:")
print(spacy_tokens)
print("Count:", len(spacy_tokens))
```

```python
#spaCy filtered tokenization

clean_tokens = [
    token.text for token in doc
    if not token.is_space and not token.is_punct
]

print("Filtered tokens:")
print(clean_tokens)
print("Count:", len(clean_tokens))
```

```python
#Stemming
from nltk.stem import PorterStemmer
import spacy

text = "Natural language processing helps computers understand human language."
nlp = spacy.load("en_core_web_sm")
doc = nlp(text)
porter = PorterStemmer()
porter_stems = [
    porter.stem(token.text.lower())
    for token in doc
    if token.is_alpha
]

print(porter_stems)
```

```python
from nltk.stem import SnowballStemmer
import spacy

text = "Natural language processing helps computers understand human language."
nlp = spacy.load("en_core_web_sm")
doc = nlp(text)
snowball = SnowballStemmer("english")

snowball_stems = [
    snowball.stem(token.text.lower())
    for token in doc
    if token.is_alpha
]

print(snowball_stems)
```

```python
#Perform Lemmatization Using spaCy

import spacy

text = "NLP enables computers and digital devices to recognize," \
" understand and generate text and speech by combining computational linguistics," \
" the rule-based modeling of human language together with statistical modeling," \
" machine learning and deep learning."
nlp = spacy.load("en_core_web_sm")
doc = nlp(text)

lemmas = [
    token.lemma_.lower()
    for token in doc
    if token.is_alpha
]

print("spaCy lemmas:")
print(lemmas)
```

```python
# Inspect POS and Lemmas Together
text = "Natural language processing helps computers understand human language."
nlp = spacy.load("en_core_web_sm")
doc = nlp(text)

for token in doc:
    if token.is_alpha:
        print(
            f"{token.text:15} "
            f"POS={token.pos_:8} "
            f"TAG={token.tag_:8} "
            f"LEMMA={token.lemma_}"
        )
```

```python

#Generate a Word-Level Comparison
import pandas as pd
import spacy
from nltk.stem import PorterStemmer
from nltk.stem import SnowballStemmer
nlp = spacy.load("en_core_web_sm")
text = "Natural language processing helps computers understand human language."
doc = nlp(text)
porter = PorterStemmer()
snowball = SnowballStemmer("english")
comparison = pd.DataFrame({
    "Original": [token.text for token in doc if token.is_alpha],
    "POS": [token.pos_ for token in doc if token.is_alpha],
    "Porter Stem": [
        porter.stem(token.text.lower())
        for token in doc if token.is_alpha
    ],
    "Snowball Stem": [
        snowball.stem(token.text.lower())
        for token in doc if token.is_alpha
    ],
    "spaCy Lemma": [
        token.lemma_.lower()
        for token in doc if token.is_alpha
    ],
})

print(comparison.to_string(index=False))
```

```python

# Unique alphabetic tokens
from nltk.stem import PorterStemmer, SnowballStemmer
import pandas as pd
import spacy

nlp = spacy.load("en_core_web_sm")
text = "NLP is a transformative technology "\
"that allows machines to understand, interpret, and generate human language,"\
"making human-computer interaction more natural and efficient."\
"Its applications span from everyday tools like virtual"\
"assistants and search engines to enterprise solutions for automation,analytics, " \
"and customer engagement. With ongoing advancements in machine learning "
"and AI, NLP continues to evolve, enabling more sophisticated and human-like "\
"communication between humans and machines "\

doc = nlp(text)

porter = PorterStemmer()
snowball = SnowballStemmer("english")

original_words = [
    token.text.lower()
    for token in doc
    if token.is_alpha
]

porter_words = [porter.stem(w) for w in original_words]
snowball_words = [snowball.stem(w) for w in original_words]

lemma_words = [
    token.lemma_.lower()
    for token in doc
    if token.is_alpha
]

metrics = pd.DataFrame({
    "Method": [
        "Original words",
        "Porter stemming",
        "Snowball stemming",
        "spaCy lemmatization"
    ],
    "Total words": [
        len(original_words),
        len(porter_words),
        len(snowball_words),
        len(lemma_words)
    ],
    "Unique forms": [
        len(set(original_words)),
        len(set(porter_words)),
        len(set(snowball_words)),
        len(set(lemma_words))
    ]
})

print(metrics)
```

```python
#Optional quantitative indicator

def reduction_percentage(original, normalized):
    original_unique = len(set(original))
    normalized_unique = len(set(normalized))
    return (1 - normalized_unique / original_unique) * 100

print(
    "Porter vocabulary reduction:",
    round(reduction_percentage(original_words, porter_words), 2),
    "%"
)

print(
    "Snowball vocabulary reduction:",
    round(reduction_percentage(original_words, snowball_words), 2),
    "%"
)

print(
    "spaCy lemma vocabulary reduction:",
    round(reduction_percentage(original_words, lemma_words), 2),
    "%"
)
```
